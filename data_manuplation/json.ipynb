```python
import json
import pandas as pd
obj= """{"name":"Ahmed","age":26,"is_married":true}"""
# using python built in library "Json" we can load json files using loads function
result= json.loads(obj)
print(result)

# python object
data = {'name': 'Tom Hanks', 'age': 67, 'oscars': 2,'has_odcar':True}
# convert python object to json
data1 = json.dumps(data)
print(data1)

# wrap dict in a list so pandas treats it as one row
dataframe = pd.DataFrame([data])
print(dataframe)



# read json file using pandas read_json
s1 = pd.read_json(r'D:\actors_100.json')
print(s1)
```
