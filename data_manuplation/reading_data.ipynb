```python
import pandas as pd

# read_csv always returns a DataFrame
df = pd.read_csv(r'D:\actors.csv')
df.head(1000)
```

```python
# Selecting a column after reading returns a Series, not a DataFrame
s1 = pd.read_csv(r'D:\actors.csv')['Name']
print(s1)
```

```python
# header=None tells pandas the file has no header row,
# so columns get assigned integer indices (0, 1, 2, ...) instead
df_without_header = pd.read_csv(r'D:\actors_no_header.csv', header=None)
df_without_header.head()
```

```python
# Use 'names' to assign custom column names to a headerless file
df_custom_names = pd.read_csv(r'D:\actors_no_header.csv', names=["Name", "Nationality", "Birth Year", "x", "y", "z"])
df_custom_names.head()
```

```python
# index_col sets a column as the row index instead of the default 0, 1, 2...
# Now you can access rows directly by actor name using .loc
df_indexed = pd.read_csv(r'D:\actors.csv', index_col='Name')
df_indexed.loc['Tom Hanks']
```

```python
# skiprows skips specific row numbers (0-indexed) when reading
df_skipped = pd.read_csv(r'D:\actors.csv', skiprows=[0, 2])
df_skipped.head()
```

```python
#nrows is only read number of rows and the rest fo the file is igonred
data = pd.read_csv(r'D:\actors_100.csv',nrows=20)

#it returns a reader object you loop over. Each iteration gives you the next chunk as a DataFrame
chunks  = pd.read_csv(r'D:\actors_100.csv',chunksize=5)
for chunk in chunks:
    print(chunk)
# NOW it reads 10,000 rows from disk
# processes them
# next iteration: previous chunk gone, next 10,000 loaded
```
