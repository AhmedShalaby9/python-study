```python
import pandas as pd

# By default pandas automatically recognizes these as NaN: ['', 'NULL', 'null', 'N/A', 'NA', 'NaN', 'nan', 'None']

# But real-world data often has custom representations for missing values, and that's where na_values comes in
# here we consider Unkown as NAN , we can add list or string
df = pd.read_csv(r'D:\actors_with_missing.csv',na_values=['Unkown'])
display(df.head(1000))

#This gives you a quick count of missing values per column so you understand your data
df.isna().sum()

```
